## when 제어문 사용
- when 제어문은 기존의 switch 문보다 강력함
- 일반적인 switch은 if 동등절에 해당하나, when 제어문은 범위 및 다양한 조건을 포함할 수 있음

```kotlin
// println("입력 스트링")
// val input : String = readln()
val data : Any = "Test"
when(data) {
     is String -> println("입력이 String임.")
     20, 30 -> println("입력 값이 20 또는 30임")
     in 1..10 -> println("입력 값이 1이상 10이하")
     else -> println("입력 값이 예상 범위를 벗어남")
}
```

```kotlin
fun series(n: Int): Double {
    var sum = 0.0
    var sign = 1.0 // 부호를 +, -, +, - 로 번갈아 바꾸기 위한 변수
    
    // 0부터 n-1까지 총 n번 반복 실행
    for (i in 0 until n) {
        // 분모의 규칙: 1, 3, 5, 7, 9 ... -> (2 * i + 1)
        val denominator = 2.0 * i + 1.0
        
        // 항을 더해줌 (부호 반영)
        sum += sign * (1.0 / denominator)
        
        // 다음 항을 위해 부호 반전
        sign = -sign
    }
    
    // x/4 = sum 이므로, 최종 x를 구하기 위해 4를 곱해서 반환합니다.
    return sum * 4.0
}

// 문제에서 요구한 100, 1000, 5000번 실행 결과 출력
println("series(100)  : ${series(100)}")
println("series(1000) : ${series(1000)}")
println("series(5000) : ${series(5000)}")
```

## 원주율을 통해서 원주율 계산하기
- pi/4 = 1 - 1/3 + 1/5 - 1/7 + ...

```kotlin
fun series(n: Int): Double {
     var x: Double = 0.0
     for(i in 1..n) {
         val denom = 2*i-1
         val sign = if (i%2 == 1) 1 else -1
         x += sign / denom.toDouble()
     }
     return x*4
}

println("Series(100) = ${Series(100)}")
println("Series(1000) = ${Series(1000)}")
println("Series(50000) = ${Series(50000)}")
```
